```python
block_layers_df.drop('id',axis=1,inplace=True)
```

```python
block_layers_df['base_name'] = block_layers_df['name'].apply(os.path.basename)
```

```python
block_layers_df
```

```python
def extract_layer_name(base_name,
                       district_name = district,
                       tehsil_name = tehsil):
    layer_name = base_name.split('_')
    layer_name.remove(district_name)
    layer_name.remove(tehsil_name)
    layer_name = '_'.join(layer_name)
    return layer_name
```

```python
block_layers_df['layer_name'] = block_layers_df['base_name'].apply(extract_layer_name)
```
