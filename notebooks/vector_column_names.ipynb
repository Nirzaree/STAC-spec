```python
def extract_layer_name(base_name,
                       district_name = district,
                       tehsil_name = tehsil):
    layer_name = base_name.split('_')
    layer_name.remove(district_name)
    layer_name.remove(tehsil_name)
    layer_name = '_'.join(layer_name)
    return layer_name
```

```python
block_layers_df['layer_name'] = block_layers_df['base_name'].apply(extract_layer_name)
```

```python
def extract_column_names(layer_path):
    layer_columns = ee.FeatureCollection(layer_path)
    layer_columns = layer_columns.first().propertyNames().getInfo()
    return layer_columns
```

```python
vector_layers_df['column_names'] = vector_layers_df['name'].apply(extract_column_names)
```
